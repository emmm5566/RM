```python
# 导入 numpy
```

```python
import numpy as np 
```

```python
# 创建矩阵
```

```python
np.identity(3)  # 创建一个3*3的单位矩阵
```

```python
np.zeros([3,3])  # 创建一个3*3的零矩阵
```

```python
np.random.rand(3,3)  # 创建一个3*3的随机矩阵
```

```python
np.asarray([1,2,3,4]).reshape(2,2)  # 从已有数组创建矩阵，asarray将Python列表转换成一维数组，reshape重塑矩阵形状
```

```python
# 判断两个矩阵是否相等
```

```python
# allclose，比较两个array是不是每一元素都相等，返回 True/False，默认在1e-05的误差范围内
```

```python
print("零矩阵和单位矩阵", np.allclose(np.zeros([3,3]), np.identity(3)))
print("单位矩阵和单位矩阵", np.allclose(np.identity(3), np.identity(3)))
```

```python
# 计算矩阵
```

```python
# 矩阵加法 np.add /减法 np.subtract
```

```python
A = np.asarray([1,2,3,4]).reshape(2,2)
B = np.identity(2)
print(np.add(A,B))
print(np.subtract(A,B))
```

```python
# 矩阵乘法 np.dot
```

```python
# （前提同形）任何矩阵乘上零矩阵等于零矩阵，任何矩阵乘上单位矩阵等于其本身
A = np.asarray([1,2,3,4]).reshape(2,2)
B = np.zeros([2,2])
C = np.identity(2)
print(np.dot(A,B))
print(np.dot(A,C))
```

```python
# 矩阵求逆 np.linalg.inv
```

```python
# 矩阵的逆与矩阵的乘积为单位矩阵
A = np.asarray([1,2,3,4]).reshape(2,2)
A_INV = np.linalg.inv(A)
print(A_INV)
print(np.dot(A,A_INV))
```

```python
# 不是所有的矩阵都有逆，比如零矩阵就没有逆，如果尝试用numpy来求逆则会出现错误
O = np.zeros(2)
# O_INV = np.linalg.inv(O)
```

```python
c = np.asarray([
    1,2,3,
    4,5,6,
    7,8,9,
]).reshape((3,3))
print("矩阵的行列式：", np.linalg.det(c))
print("矩阵的秩：", np.linalg.matrix_rank(c))
c_inv = np.linalg.inv(c)
print("矩阵c和c_inv点乘的结果：\n", np.dot(c,c_inv))
```
